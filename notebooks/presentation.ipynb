Hello, my name is Regina and last month I started working at Fit.ly Tech as a Data Analyst. It is really nice to meet you for the first time and I am pleased to share my findings on drivers of customer churn during the last 6 month here at Fit.ly Tech. 

As the head of data analytics told me, it is very crucial to reduce the churn because on the one hand costs for acquiring new users rise and on the other hand we want to keep the existing users so that the revenue stays stable and so that the pressure on Marketing and Product departments does not have to rise. So lets dive in!

On this slide you can see a small summary of the data i worked with. The analyzed time period of the datasets was 6 month and I got three datasets from different departments of the company. 
The first data set was general user information, mainly the account plan, the price paid and if the account is already churned. 
The second one was information about customer support tickets, mainly about when the ticket was opened and how long it needed to be solved, as well as what the topic of the ticket was and through which communication channel is came to the company. 
The third dataset was about user activities, mainly about when the user was active and what he does.
So to give some total numbers, there is a total of 400 subscribed users at Fit.ly Tech, but yet the data analysis could only include data of 357 users because the others did not consent the GDPR. As you can see on the right, from those 357 analysed users, 97 already churned, whereas 250 users still have active accounts.

Let’s take a closer look at the potential drivers of churn. I first explored the plan type the users subcribed to. As you can see on the left, there are 94 users with a free plan whereas 263 users have a paid plan. On the right you can see that almost 40% of the free users already churned, whereas only around 20% churned on paid plans. So we can conclude that the plan type may be associated with churn.

Here you can see the relationship between churn and customer support resolution time of ticket. Please keep in mind that not all users opened tickets, so these visualizations only cover users with tickets. On the left you can see that the resolution time ranges between 1 hour and 32 hours, with a median resolution time of 7 hours. On the right you can see that the resolution time may be a potential driver of customer churn, as median and range are higher when looking at churned accounts compared with non-churned accounts.

If we look at the relationship between churn and user activity information, you can see on the left, that approximatly 135 users never did an activity on the platform and the churn with no activity is higher than it is if a user did one or more activity. 

So, what can the business do to monitor the churn and take action to reduce it. I would suggest to monitor the churn rate which is calculated as the number of churned accounts divided by the number of total accounts. Or maybe if we take into account that we only generate revenue with paid accounts, we can instead monitor the churn rate of users with paid plans. 
In my analysis I used the data of a 6-month period, which resulted in a total churn rate of 27%. Thus for future monitoring, I would suggest to calculate the churn rate every month or every quarter. With that, we can react more quickly if churn rates tend to increase.

To conclude, I would suggest to start monitoring the churn rate, and maybe also the churn rate for paid plans on a monthly or quarterly basis. For future analysis, it is crucial that also the start date of the account and the end date or churn date of the account is part of the dataset to analyse. So I think the engineering team should check if these data points are already part of the database and if not, add it to the database.
As the analysis also associated a relationship between churn and customer support tickets i would suggest to review the support process to maybe find some things to improve there.
Concerning user activity on the platform I would suggest to look for ways to engage user activities as the users get a better insight of the functionalities of the platform and maybe change also from free to paid plan instead of thinking about churn.
